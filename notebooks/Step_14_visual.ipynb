```python
import pandas as pd

kpi_df   = pd.read_csv("data/processed/kpi_results.csv",       index_col=0)
hist_var = pd.read_csv("data/processed/historical_var.csv",     index_col=0)
para_var = pd.read_csv("data/processed/parametric_var.csv",     index_col=0)
mc_var   = pd.read_csv("data/processed/monte_carlo_var.csv",    index_col=0)
clusters = pd.read_csv("data/processed/clustering_results.csv", index_col=0)
backtest = pd.read_csv("data/processed/backtest_results.csv",   index_col=0)

sectors_map = {
    "JPM": "Financials", "GS": "Financials",  "HLI": "Financials",
    "AAPL": "Tech",      "PLTR": "Tech",       "META": "Tech",
    "JNJ": "Healthcare", "MRK": "Healthcare",  "UNH": "Healthcare",
    "MCD": "Staples",    "KO": "Staples",
    "EL": "Beauty",      "ULTA": "Beauty",
    "XOM": "Energy",     "CVX": "Energy"
}

kpi_df["Sector"] = kpi_df.index.map(sectors_map)

var_combined = pd.DataFrame({
    "Historical_95%": hist_var["VaR_95%"] * 100,
    "Parametric_95%": para_var["VaR_95%"] * 100,
    "MonteCarlo_95%": mc_var["VaR_95%"]   * 100,
    "Historical_99%": hist_var["VaR_99%"] * 100,
    "Parametric_99%": para_var["VaR_99%"] * 100,
    "MonteCarlo_99%": mc_var["VaR_99%"]   * 100,
})
var_combined["Sector"]      = var_combined.index.map(sectors_map)
var_combined["Avg_VaR_95%"] = (
    var_combined["Historical_95%"] +
    var_combined["Parametric_95%"] +
    var_combined["MonteCarlo_95%"]
) / 3

with pd.ExcelWriter("reports/powerbi_data.xlsx") as writer:
    kpi_df.to_excel(writer,       sheet_name="KPIs")
    var_combined.to_excel(writer,  sheet_name="VaR Results")
    clusters.to_excel(writer,      sheet_name="Clustering")
    backtest.to_excel(writer,      sheet_name="Backtesting")

print("Excel saved to reports/powerbi_data.xlsx ✅")
print("\nSheets created:")
print("  ✅ KPIs")
print("  ✅ VaR Results")
print("  ✅ Clustering")
print("  ✅ Backtesting")
```
